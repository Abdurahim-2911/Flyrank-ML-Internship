## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

1. When we take a look at the freshness tier which has the highest impressions, ctr and other factors, 181+ seems like the perfect choice to pursue further. Given that the internet trends also make the assumption that the best option is to update constantly. But this insight is decieving. When we look at the total number of unique websites that have the freshness tier of 181+, it is *174* which is the lowest among all. Keeping that in mind when look at the acutal percentage of ctr per insight we see that 181+ actually ranks lower than both 0-30 and 91-80. This shows that the best option here is not what we previously thought.

2. In the second code snipped we can observe a similar insight. The avg of ctr, impressions and other factors is the highest for the websites with lowest word count *<1000* . But we can see that the number of unique websites with word count tier lower than 1000 is the lowest *(973)*. So if we look at a large sample with better combined avg ctr per impression. The websites with 2000-3500 is a better option.

```python
# RUN THIS CELL FIRST AND THEN EITHER OF THE OTHER TWO
import pandas as pd
url = 'https://raw.githubusercontent.com/Abdurahim-2911/Flyrank-ML-Internship/refs/heads/main/data/raw/content_refresh_anonymized.csv'
df = pd.read_csv(url)

```

```python
print(df.groupby(['freshness_tier'])['content_id'].nunique())
print(df.groupby(['freshness_tier'])[['impressions_90d','ctr','scroll_rate','avg_position']].mean())
```

```python
print(df.groupby(['word_count_tier'])['content_id'].nunique())
print(df.groupby(['word_count_tier'])[['impressions_90d','ctr','scroll_rate','avg_position']].mean())
```
